```python
#create random images
import matplotlib.pyplot as plt

# 1. Create your example 4D tensor (Batch, Channels, Height, Width)
tensor_4d = torch.rand(1, 3, 64, 64)

# 2. Remove the batch dimension -> shape becomes (3, 64, 64)
tensor_3d = tensor_4d.squeeze(0)  # alternative: tensor_4d[0]

# 3. Rearrange channels from (C, H, W) to (H, W, C) for matplotlib
display_img = tensor_3d.permute(1, 2, 0)

# 4. Plot the image
plt.imshow(display_img.numpy())
plt.axis('off')  # Hide pixel grid coordinates
plt.show()
```

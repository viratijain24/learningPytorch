```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return 3.14 * self.radius**2
```

```python
print(vars(Circle))
```

```python
circle1 = Circle(5)
print(vars(circle1))
print(type(circle1))
print(circle1.radius)
print(circle1.calculate_area())
print(Circle.calculate_area(circle1))
```

```python
circle1.radius = 7
print(circle1.radius)
```

naming conventions:

class names start with capital letter, and capitalizes each word instead of using underscores

Member	      /Naming	                               /Examples

Public	      /Use the normal naming pattern.	       /radius, calculate_area()

Non-public	  /Include a leading underscore in names.   /_radius, _calculate_area()

if __ is added to name of any attribute/ method, then its name is mangled, the name of the class is prepended to the name automatically

name       → public
_name      → non-public convention
__name     → name mangling
__name__   → usually special/dunder method

Class Attributes vs Instance Attributes

Class Attributes are used in the class, outside of any method. all instances created have it as an attribute

eg: class ObjectCounter:

...     num_instances = 0

...     def __init__(self):

...         ObjectCounter.num_instances += 1
or type(self).num_instances += 1

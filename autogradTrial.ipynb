```python
import torch

a = torch.tensor([2., 3.], requires_grad=True)
b = torch.tensor([6., 4.], requires_grad=True)
```

```python
Q = 3*a**3 - b**2
```

```python
external_grad = torch.tensor([1., 1.])
Q.backward(gradient=external_grad)
```

```python
# check if collected gradients are correct
print(9*a**2 == a.grad)
print(-2*b == b.grad)
```

```python
w = torch.tensor(2.0, requires_grad=True)   # tell autograd to track this tensor
x = torch.tensor(3.0, requires_grad=True)
y = torch.tensor(10.0)

pred = w * x
loss = (pred - y) ** 2

loss.backward()          # walks the computational graph backward, computes gradients
print(w.grad)              # dL/dw — the gradient PyTorch computed automatically
print(x.grad)              #dL/dx
print(y.grad)
```
